# Как устроен Jupyter Notebook

- Любой ноутбук состоит из **ячеек** (cells). Даже этот текст написан в ячейке.

## Типы ячеек
1. **Code** — код на Python, выполняется при запуске.
2. **Markdown** (текст) — заметки, заголовки, формулы в LaTeX.

Тип ячейки переключается через верхнее меню или горячими клавишами: `Y` — в код, `M` — в текст (в командном режиме).

## Два режима
- **Режим редактирования** — вы печатаете внутри ячейки. Вход: `Enter` или клик внутрь ячейки.
- **Командный режим** — вы управляете ячейками как объектами. Вход: `Esc`.

## Полезные горячие клавиши (командный режим)
| Клавиша | Действие |
|---|---|
| `Shift + Enter` | выполнить ячейку и перейти к следующей |
| `Ctrl + Enter` | выполнить ячейку и остаться в ней |
| `A` / `B` | вставить ячейку выше / ниже |
| `D, D` (дважды) | удалить ячейку |
| `Z` | отменить удаление |

## Как остановить выполнение
Иногда ячейку нужно прервать (например, бесконечный цикл):
- кнопка ■ (Stop) на панели инструментов;
- меню *Kernel → Interrupt* (в Colab: *Среда выполнения → Прервать выполнение*);
- радикально: *Kernel → Restart* — перезапуск ядра, все переменные будут потеряны.

У списков есть много **методов** (функций, специфичных именно для списков):

- **`list.append(x)`** — добавить элемент `x` в конец списка.
- **`list.extend(L)`** — расширить список, добавив в конец все элементы из списка `L`.
- **`list.insert(i, x)`** — вставить элемент `x` на позицию `i`.
- **`list.remove(x)`** — удалить первый элемент, равный `x`. Если такого нет — `ValueError`.
- **`list.pop([i])`** — удалить элемент на позиции `i` и вернуть его. Без аргумента удаляет и возвращает последний элемент.
- **`list.index(x)`** — вернуть индекс первого элемента, равного `x`. Если такого нет — `ValueError`.
- **`list.count(x)`** — сколько раз `x` встречается в списке.
- **`list.sort(key=...)`** — отсортировать список на месте.
- **`list.reverse()`** — развернуть список на месте.
- **`list.copy()`** — вернуть поверхностную копию списка.
- **`list.clear()`** — удалить все элементы.

### Логические выражения

Логическое выражение может быть истинным или ложным. Простейший вид:

`<арифметическое выражение> <знак сравнения> <арифметическое выражение>`

Например, если есть переменные `x` и `y`, то в выражении `x + y < 3*y` слева стоит `x + y`, знак сравнения — `<` (меньше), справа — `3*y`.

<center>
    <table>
        <tr>
            <th>Оператор</th>
            <th>Значение</th>
        </tr>
        <tr><td>&lt;</td><td>меньше</td></tr>
        <tr><td>&gt;</td><td>больше</td></tr>
        <tr><td>&lt;=</td><td>меньше или равно</td></tr>
        <tr><td>&gt;=</td><td>больше или равно</td></tr>
        <tr><td>==</td><td>равно (не путать с присваиванием <code>=</code>!)</td></tr>
        <tr><td>!=</td><td>не равно</td></tr>
    </table>
</center>

````python
x = 1 > 2
print(type(x))
print(x)
print(int(x))   # False превращается в 0, True — в 1
````

Операции `and` и `or` — бинарные, т.е. пишутся между операндами: `x < 3 or y > 2`.
Операция `not` — унарная, пишется перед единственным операндом: `not x`.

````python
x = int(input('x = '))
y = int(input('y = '))
z = int(input('z = '))
print(x == y)
print(x > y and y < z)
````

### Условный оператор

Чаще всего логические выражения используются в условных операторах.

Синтаксис:

```python
if условие_1:
    команды
elif условие_2:
    команды
elif условие_3:
    команды
else:
    команды
```

Обратите внимание на **отступы** (4 пробела): именно они показывают Python, какие команды относятся к какой ветке.

````python
x = 5
y = 7
if x > y:
    print('X больше Y')
    print(x - y)
else:
    print(' ')
    print('X меньше или равен Y')
````

````python
x = int(input('Введите X: '))
y = int(input('Введите Y: '))
if x > y:
    print('X больше Y')
    print(x - y)
elif y > x:
    print('Y больше X')
    print(y - x)
else:
    print('X равен Y')
````

### Цикл `while`

Как только строка кода выполнена, она «забывается»; чтобы повторить действие, его пришлось бы писать заново. Для многократного повторения одних и тех же действий используются **циклы**.

**`while`** выполняет блок команд, пока условие истинно. После выполнения блока управление возвращается к строке с условием: если оно всё ещё верно — блок повторяется, если нет — выполнение продолжается после цикла.

```python
while условие:
    команды
```

Важно случайно не зациклиться навсегда! Если это произошло — нажмите ■ (Stop).

````python
temperature = 5
while temperature > 0:
    print('Погода нормальная.')
    temperature = temperature - 1
````

## Функции

Если какие-то действия приходится повторять снова и снова, а встроенной функции нет — можно написать свою.

Синтаксис:
```python
def имя_функции(аргументы):
    команды
    return результат
```

### Векторные операции (вычисления без циклов)

#### Пример: поэлементное сложение двух одномерных массивов

````python
import numpy as np

# фиксируем генератор случайных чисел, чтобы результаты воспроизводились
np.random.seed(42)

# создаём два NumPy-массива
a_numpy = np.random.randint(low=1, high=10, size=5)
b_numpy = np.random.randint(low=1, high=10, size=5)
print(f'a_numpy: {a_numpy}, b_numpy: {b_numpy}, type: {type(a_numpy)}')

# и два аналогичных списка
a_list = a_numpy.tolist()
b_list = b_numpy.tolist()
print(f'a_list: {a_list}, b_list: {b_list}, type: {type(a_list)}')
````

> **Примечание.** `np.random.seed()` — «старый» API, он по-прежнему работает. В современном NumPy рекомендуется создавать отдельный генератор:
> ```python
> rng = np.random.default_rng(42)
> rng.integers(low=1, high=10, size=5)
> ```
> Так у каждого куска кода свой независимый источник случайности, и они не мешают друг другу.

Чистый Python:

````python
sum_arr = []
for i in range(len(a_list)):
    sum_arr.append(a_list[i] + b_list[i])
print(sum_arr)
````

Векторизованное решение с NumPy:

````python
sum_numpy = a_numpy + b_numpy
print(sum_numpy)
````

````python
%%timeit
sum_arr = []
for i in range(len(a_list)):
    sum_arr.append(a_list[i] + b_list[i])
````

````python
%%timeit
sum_numpy = a_numpy + b_numpy
````

На массивах из 5 элементов разница невелика (а иногда NumPy даже проигрывает — есть накладные расходы на вызов). Попробуйте увеличить `size` до 1 000 000 и сравнить снова — разрыв станет в сотни раз.

#### Произведение двух матриц

````python
import random
````

````python
# ещё один способ зафиксировать случайность — для стандартного модуля random
random.seed(42)
````

````python
np.random.seed(42)

# две матрицы
a_numpy = np.random.randint(low=1, high=10, size=(3, 4))
b_numpy = np.random.randint(low=1, high=10, size=(4, 3))
print(f'a_numpy:\n{a_numpy}\n\nb_numpy:\n{b_numpy}\n\ntype: {type(a_numpy)}')

# и два аналогичных списка списков
a_list = a_numpy.tolist()
b_list = b_numpy.tolist()
print(f'\na_list:\n{a_list}\n\nb_list:\n{b_list}\n\ntype: {type(a_list)}')
````

* Чистый Python — три вложенных цикла:

````python
prod_arr = [[], [], []]

for i in range(len(a_list)):
    for j in range(len(a_list)):
        sum_ = 0
        for k in range(len(b_list)):
            sum_ += a_list[i][k] * b_list[k][j]
        prod_arr[i].append(sum_)
````

````python
print(prod_arr)
````

* Те же циклы, но по NumPy-массиву. **Внимание:** это *не* векторизация — поэлементный доступ к `ndarray` в цикле медленнее, чем к списку:

````python
c = np.zeros((a_numpy.shape[0], b_numpy.shape[1]))
for i in range(a_numpy.shape[0]):
    for j in range(b_numpy.shape[1]):
        for s in range(a_numpy.shape[1]):
            c[i, j] += a_numpy[i, s] * b_numpy[s, j]
````

````python
print(c)
````

* Встроенное матричное умножение. Оператор `@` и метод `.dot()` делают одно и то же; `@` читается лучше:

````python
a_numpy @ b_numpy
````

````python
%%timeit
prod_arr = [[], [], []]

for i in range(len(a_list)):
    for j in range(len(a_list)):
        sum_ = 0
        for k in range(len(b_list)):
            sum_ += a_list[i][k] * b_list[k][j]
        prod_arr[i].append(sum_)
````

````python
%%timeit
c = np.zeros((a_numpy.shape[0], b_numpy.shape[1]))
for i in range(a_numpy.shape[0]):
    for j in range(b_numpy.shape[1]):
        for s in range(a_numpy.shape[1]):
            c[i, j] += a_numpy[i, s] * b_numpy[s, j]
````

````python
%%timeit
a_numpy @ b_numpy
````

#### Пример: цепочка вычислений

Дана функция

$f(x) = \frac{1}{2}x^TAx + b^Tx + c$

и её градиент

$\nabla{f(x)} = \frac{1}{2}(A + A^T)x + b$

Вычислим $f(x)$ и $\nabla{f(x)}$ в заданной точке $x$.

````python
np.random.seed(42)
A = np.random.randint(low=1, high=10, size=(3, 3))
print(A)
````

````python
x = np.random.randint(1, 10, 3)
b = np.random.randint(1, 10, 3)
c = np.random.randint(1, 10, 1)
print('x:', x)
print('b:', b)
print('c:', c)
````

````python
0.5 * x @ A @ x + b @ x + c
````

````python
A
````

````python
0.5 * (A + A.T) @ x + b
````

> **Тонкость:** для одномерного массива `x.T` ничего не делает — у вектора формы `(3,)` нет «строк» и «столбцов». NumPy сам разбирается, как умножать вектор на матрицу слева (`x @ A`) и справа (`A @ x`). Транспонирование имеет смысл только для двумерных массивов.

````python
x @ A
````

````python
x.T @ A     # то же самое
````

Сгруппируем данные: для пяти стран с наибольшим числом обзоров посчитаем, сколько вин получило каждую оценку. Получится таблица «оценка × страна» — идеальный формат для `plot`, где каждый столбец станет отдельной линией.

> Здесь мы заранее используем `groupby` и `unstack` — подробно они будут разобраны позже. Пока просто посмотрите на результат.

````python
top5 = data['country'].value_counts().head(5).index

wine_counts = (
    data[data['country'].isin(top5)]
    .groupby(['points', 'country'])
    .size()
    .unstack('country', fill_value=0)
)
wine_counts.head()
````

````python
wine_counts.plot.line(figsize=(10, 5))
````

````python
wine_counts.plot.area(figsize=(10, 5))
````

````python
wine_counts.plot.bar(stacked=True, figsize=(12, 5))
````

# Ещё немного визуализации, или введение в Matplotlib

`df.plot()` — это удобная обёртка над **Matplotlib**. Когда нужен полный контроль над графиком (подписи, цвета, несколько осей), обращаемся к самой библиотеке.

````python
import matplotlib.pyplot as plt
````

График «по умолчанию»:

````python
# немного синтетических данных
import numpy as np
x = np.linspace(0, 15, 100)

# график по умолчанию: по оси X — номера элементов
plt.plot(x)
plt.show()
````

````python
# график по умолчанию №2
y = np.cos(x)
plt.plot(y)
plt.show()
````

Что не так с графиком выше?

* нет подписей осей;
* нет заголовка;
* по оси X — номер точки, а не значение `x`.

Исправим:

````python
plt.plot(x, y)
plt.title('cos(x)')
plt.xlabel('x')
plt.ylabel('y')
plt.show()
````

Добавим легенду — для этого у каждой линии должен быть `label`:

````python
z = 0.5 * np.sin(x)

plt.plot(x, y, label='cos(x)')
plt.plot(x, z, label='0.5 sin(x)')
plt.title('0.5 sin(x) и cos(x)')
plt.xlabel('x')
plt.ylabel('y, z')
plt.legend()
plt.show()
````

Всё равно тесновато. Изменим размер фигуры:

````python
plt.figure(figsize=(8, 4))
plt.plot(x, y, label='cos(x)')
plt.plot(x, z, label='0.5 sin(x)')
plt.title('0.5 sin(x) и cos(x)')
plt.xlabel('x')
plt.ylabel('y, z')
plt.legend()
plt.show()
````

Как поменять цвета? Аргумент `color` (или коротко `c`):

````python
plt.figure(figsize=(15, 6))
plt.plot(x, y, label='cos(x)', color='g')
plt.plot(x, z, label='0.5 sin(x)', color='r')
plt.title('0.5 sin(x) и cos(x)')
plt.xlabel('x')
plt.ylabel('y, z')
plt.legend()
plt.show()
````

### Несколько графиков на одной фигуре

Современный способ — `plt.subplots(nrows, ncols)`: он возвращает фигуру и массив осей, с каждой из которых работаем отдельно. Аргумент `sharex=True` связывает оси X.

````python
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(10, 5), sharex=True)

ax1.plot(x, y, 'b-.', label='cos(x)')
ax1.set_title('Два графика с общей осью X')
ax1.legend()

ax2.plot(x, z, 'r--', label='0.5 sin(x)')
ax2.set_xlabel('Общая ось x')
ax2.legend()

plt.show()
````

> Строка `'b-.'` — короткая запись стиля: `b` — синий цвет, `-.` — штрихпунктирная линия. `'r--'` — красная пунктирная.
